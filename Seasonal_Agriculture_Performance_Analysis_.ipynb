# Major Project: Seasonal Agriculture Performance Analysis

```python

# # 1. Import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# # 2. Upload Dataset

df = pd.read_csv("seasonal_agriculture_performance_dataset.csv")
df.head()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
# Step 6: Clean Missing Values

df['Rainfall_mm'] = df['Rainfall_mm'].fillna(df['Rainfall_mm'].median())
df['Soil_Moisture_pct'] = df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median())
df['Yield_Tonnes_Ha'] = df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median())

print("Missing values after cleaning:")
print(df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
```

```python

print("Duplicate rows:", df.duplicated().sum())
```

```python
# remove duplicates if any
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

```python
df['Season'].value_counts()
```

```python
df['Season'].value_counts().plot(kind='bar')
plt.title("Number of Records per Season")
plt.xlabel("Season")
plt.ylabel("Count")
plt.show()
```

```python
avg_yield = df.groupby('Season')['Yield_Tonnes_Ha'].mean()
print(avg_yield)
```

```python
avg_yield.plot(kind='bar', color='green')
plt.title("Average Yield (Tonnes/Ha) by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield")
plt.show()
```

```python
# Step 8: Handle Missing Values

columns_to_fill = [
    "Rainfall_mm",
    "Soil_Moisture_pct",
    "Yield_Tonnes_Ha"
]

for col in columns_to_fill:
    if col in df.columns:
        df[col] = df.groupby("Season")[col].transform(
            lambda x: x.fillna(x.median())
        )

# Remaining missing values ko overall median se fill karna
for col in columns_to_fill:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].median())

print("Missing values after cleaning:")
print(df.isnull().sum())
```

```python
# Step 11: Research Questions

questions = [
    "1. Agricultural performance different seasons mein kaise change hoti hai?",
    "2. Kis season mein average crop yield sabse zyada hai?",
    "3. Kis season mein average profit sabse zyada hai?",
    "4. Seasons ke beech rainfall aur temperature mein kya difference hai?",
    "5. Water usage aur water efficiency season ke according kaise change hoti hai?",
    "6. Different crops ki performance seasons ke across kaise change hoti hai?",
    "7. Irrigation methods ka yield par kya difference hai?",
    "8. Disease aur pest risk ka yield ke saath kya relationship hai?",
    "9. Environmental factors aur crop yield ke beech kya relationship hai?",
    "10. Kya different seasons ki yield mein statistically significant difference hai?"
]

for q in questions:
    print(q)
```

```python

crop_season_yield = df.groupby(['Crop','Season'])['Yield_Tonnes_Ha'].mean().unstack()
crop_season_yield
```

```python

crop_season_yield.plot(kind='bar', figsize=(10,6))
plt.title("Average Yield by Crop and Season")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xticks(rotation=30)
plt.show()
```

```python
# ## Step 12: Which Irrigation Method Gives Better Yield

irrigation_yield = df.groupby('Irrigation_Method')['Yield_Tonnes_Ha'].mean()
print(irrigation_yield)
```

```python
irrigation_yield.plot(kind='bar', color='purple')
plt.title("Average Yield by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Average Yield")
plt.show()
```

```python

# ## Step 13: Disease / Pest Risk by Season

disease_risk = df.groupby('Season')['Disease_Pest_Risk_pct'].mean()
print(disease_risk)
```

```python

disease_risk.plot(kind='bar', color='red')
plt.title("Average Disease/Pest Risk % by Season")
plt.xlabel("Season")
plt.ylabel("Risk %")
plt.show()
```

```python
# Step 14: Water Usage per Hectare

df["Water_per_Hectare_m3"] = np.where(
    df["Farm_Area_Hectares"] > 0,
    df["Water_Used_m3"] / df["Farm_Area_Hectares"],
    np.nan
)

water_per_hectare = (
    df.groupby("Season")["Water_per_Hectare_m3"]
    .mean()
    .sort_values(ascending=False)
)

print("Average Water Used per Hectare:")
print(water_per_hectare)

plt.figure(figsize=(8, 5))
water_per_hectare.plot(kind="bar")

plt.title("Average Water Usage per Hectare by Season")
plt.xlabel("Season")
plt.ylabel("Water Used per Hectare (m³)")
plt.xticks(rotation=0)

plt.show()
```

```python
# Check important numeric columns for impossible negative values
numeric_check_cols = [
    'Farm_Area_Hectares', 'Rainfall_mm', 'Humidity_pct',
    'Soil_Moisture_pct', 'Yield_Tonnes_Ha',
    'Water_Used_m3', 'Disease_Pest_Risk_pct'
]

for col in numeric_check_cols:
    negative_count = (df[col] < 0).sum()
    print(f"{col}: negative values = {negative_count}")

# Check percentage ranges
for col in ['Humidity_pct', 'Soil_Moisture_pct', 'Disease_Pest_Risk_pct']:
    print(f"{col}: min={df[col].min():.2f}, max={df[col].max():.2f}")

# Show State-District combinations for manual data-quality review
state_district = df[['State', 'District']].drop_duplicates()
print("\nNumber of unique State-District combinations:", len(state_district))
state_district.head(20)
```

```python
# Step 15: Correlation Analysis

factor_cols = [
    "Farm_Area_Hectares",
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Soil_Moisture_pct",
    "Sunlight_Hours_Day",
    "Water_Used_m3",
    "Water_Efficiency_t_per_1000m3",
    "Disease_Pest_Risk_pct",
    "Yield_Tonnes_Ha"
]

available_factor_cols = [
    col for col in factor_cols if col in df.columns
]

correlation = df[available_factor_cols].corr()

print("Correlation with Yield:")
print(correlation["Yield_Tonnes_Ha"].sort_values(ascending=False))

plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Between Agricultural Factors and Yield")
plt.show()
```

```python
# # Step 16: Yield Distribution and Outliers
# Boxplots help identify unusual yield and profit observations.
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Season', y='Yield_Tonnes_Ha')
plt.title("Yield Distribution and Outliers by Season")
plt.xlabel("Season")
plt.ylabel("Yield (Tonnes/Ha)")
plt.show()

plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='Season', y='Profit_INR')
plt.title("Profit Distribution and Outliers by Season")
plt.xlabel("Season")
plt.ylabel("Profit (INR)")
plt.show()
```

```python
# # Step 17: Crop-wise Seasonal Change
# This section calculates the percentage difference between Kharif and Zaid yield for each crop.

crop_season = df.groupby(
    ['Crop', 'Season']
)['Yield_Tonnes_Ha'].mean().unstack()

crop_season['Kharif_vs_Zaid_%'] = (
    (crop_season['Kharif'] - crop_season['Zaid'])
    / crop_season['Zaid']
) * 100

crop_season.sort_values(
    'Kharif_vs_Zaid_%',
    ascending=False
).round(2)
```

```python
# Step 18: Key Findings

season_summary = df.groupby("Season").agg({
    "Yield_Tonnes_Ha": "mean",
    "Profit_INR": "mean",
    "Rainfall_mm": "mean",
    "Avg_Temperature_C": "mean",
    "Water_Used_m3": "mean",
    "Water_Efficiency_t_per_1000m3": "mean",
    "Disease_Pest_Risk_pct": "mean"
}).round(2)

best_yield_season = season_summary["Yield_Tonnes_Ha"].idxmax()
best_profit_season = season_summary["Profit_INR"].idxmax()
best_water_season = season_summary[
    "Water_Efficiency_t_per_1000m3"
].idxmax()

highest_rainfall = season_summary["Rainfall_mm"].idxmax()
highest_risk = season_summary["Disease_Pest_Risk_pct"].idxmax()

print("KEY FINDINGS")
print("=" * 50)

print("1. Highest average yield season:", best_yield_season)
print("2. Highest average profit season:", best_profit_season)
print("3. Best water efficiency season:", best_water_season)
print("4. Highest rainfall season:", highest_rainfall)
print("5. Highest disease/pest risk season:", highest_risk)
```

```python
# Step 19: Practical Recommendations

print("PRACTICAL RECOMMENDATIONS")
print("=" * 50)

print("1. High water-use seasons mein water usage ko carefully monitor karna chahiye.")
print("2. Efficient irrigation methods ka use water saving ke liye kiya ja sakta hai.")
print("3. High disease/pest risk seasons mein crop monitoring badhani chahiye.")
print("4. Crop planning mein rainfall, temperature aur soil moisture ko consider karna chahiye.")
print("5. Farmers seasonal yield aur profit compare karke suitable crops select kar sakte hain.")
```

```python

# # Step 20: Economic Risk Analysis

loss_rate = (
    df.assign(Loss=df['Profit_INR'] < 0)
      .groupby('Season')['Loss']
      .mean() * 100
)

print("Loss-making farms (%):")
print(loss_rate.round(2))

loss_rate.plot(kind='bar', figsize=(8, 5))
plt.title("Percentage of Loss-Making Farms by Season")
plt.xlabel("Season")
plt.ylabel("Loss-Making Farms (%)")
plt.tight_layout()
plt.show()
```

```python

# # Step 21: Profit Margin

df['Profit_Margin_pct'] = np.where(
    df['Revenue_INR'] != 0,
    (df['Profit_INR'] / df['Revenue_INR']) * 100,
    np.nan
)

profit_margin = df.groupby('Season')[
    'Profit_Margin_pct'
].mean().round(2)

print("Average Profit Margin (%):")
print(profit_margin)
```

```python
# Step 22: Regional Analysis

state_season_yield = df.groupby(
    ['State', 'Season']
)['Yield_Tonnes_Ha'].mean().unstack()

print("Average Yield by State and Season:")
print(state_season_yield.round(2))

state_season_yield.plot(
    kind='bar',
    figsize=(12, 6)
)
plt.title("Average Yield by State and Season")
plt.xlabel("State")
plt.ylabel("Yield (Tonnes/Ha)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

state_season_profit = df.groupby(
    ['State', 'Season']
)['Profit_INR'].mean().unstack()

print("Average Profit by State and Season:")
print(state_season_profit.round(2))

state_season_profit.plot(
    kind='bar',
    figsize=(12, 6)
)
plt.title("Average Profit by State and Season")
plt.xlabel("State")
plt.ylabel("Average Profit (INR)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

# # Step 23: Season-wise Environmental Relationships

environment_cols = [
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Soil_Moisture_pct',
    'Sunlight_Hours_Day',
    'Disease_Pest_Risk_pct'
]

season_corr = {}

for season in df['Season'].dropna().unique():
    season_data = df[df['Season'] == season]
    season_corr[season] = (
        season_data[environment_cols + ['Yield_Tonnes_Ha']]
        .corr()['Yield_Tonnes_Ha']
        .drop('Yield_Tonnes_Ha')
        .sort_values(ascending=False)
    )

for season, values in season_corr.items():
    print("\n", "=" * 50)
    print(season)
    print(values.round(3))

```

```python
# Step 24 : Correlation Between Yield and Agricultural Factors

factor_cols = [
    'Farm_Area_Hectares',
    'Rainfall_mm',
    'Avg_Temperature_C',
    'Humidity_pct',
    'Soil_Moisture_pct',
    'Sunlight_Hours_Day',
    'Water_Used_m3',
    'Water_Efficiency_t_per_1000m3',
    'Disease_Pest_Risk_pct'
]

correlation = df[factor_cols + ['Yield_Tonnes_Ha']].corr()

plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0
)

plt.title("Correlation Between Agricultural Factors and Yield")
plt.tight_layout()
plt.show()
```
